# Analytical model of a hybrid reluctance actuator

## 1 Electromechanical dynamics

```{figure} ./actuator_dimension.svg
Sketch showing the structure of a hybrid reluctance actuator.
```



### 1.1 Assumptions

* Magnetic core is infinitely permeable
* The permanent magnet height is far larger than the lengths of all air-gaps
* Ignoring all leakages and fringing effect.


### 1.2 Electrical dynamics

Voltage equation:

$$
u(t) = Ri(t) + \frac{\mathrm{d}\lambda(x,i)}{\mathrm{d}t},
$$
where $u(t)$ is the input voltage, $i$ is the coil current $\lambda$ is the coil flux linkage.

Flux linkage equation:

$$
\lambda (x,i) = \lambda_m(x) + \lambda_c(i),
$$
where $\lambda_m(x)$ is the flux linakge contributed by the permanent magnet, which is independent from current because saturation is neglected. $\lambda_c(i)$ is the flux linkage contributed by the coil current only. Since we assumed there is no leakage and saturation, the coil inductance is considered to be a constant $L$,

$$
\lambda_c(i) = Li(t).
$$

Substituting it to the voltage equation,

$$
\begin{aligned}
u(t) &= Ri(t) + \frac{\mathrm{\partial}\lambda(x,i)}{\mathrm{\partial}x}\frac{\mathrm{d}x}{\mathrm{d}t}+\frac{\mathrm{\partial}\lambda(x,i)}{\mathrm{\partial}i}\frac{\mathrm{d}i}{\mathrm{d}t}\\
     &= Ri(t) + v\frac{\mathrm{\partial}\lambda(x,i)}{\mathrm{\partial}x} + L\frac{\mathrm{d}i}{\mathrm{d}t},
\end{aligned}
$$
where the first item on the right represent the resistive voltage drop, the second the motional voltage (back-emf) and the third the transformer voltage.

Multiplied by $i(t)$ on both sides, the poewr balancing equation is obtained

$$
\begin{aligned}
u(t)i(t) &= Ri^2(t) + i(t)v(t)\frac{\mathrm{\partial}\lambda(x,i)}{\mathrm{\partial}x} + i(t)L\frac{\mathrm{d}i}{\mathrm{d}t},
\end{aligned}
$$
where the first item on the right represent the Ohmic loss, the second the mechanical power and the third the change of magnetic energy stored in the inductor.

Therefore the mechanical force on the mover is solved from power balancing by

$$
f(x,i) = \frac{i(t)v(t)\frac{\mathrm{\partial}\lambda(x,i)}{\mathrm{\partial}x}}{v(t)} = i(t)\frac{\mathrm{\partial}\lambda(x,i)}{\mathrm{\partial}x}.
$$

### 1.3 Mechanical dynamics

According to Newton's law,

$$
f(x,i) - kx = m\frac{\mathrm{d}v(t)}{\mathrm{d}t} = m\frac{\mathrm{d}^2x(t)}{\mathrm{d}t^2},
$$
where $k$ is the stiffness of the flexure to counteract the negative stiffness of the actuator.

## 2 Magnetic circuit analysis

```{figure} ./actuator_magnetic_circuit.svg

Magnetic circuit of the investigated hybrid reluctance actuator.
```

### 2.1 Coil excitation

The number of turns of the coil on each side is $N$. The two sides connected in series, resulting a total number of turns in series of $2N$.
The flux linkaged by the coil goes through the two horizontal air gaps $g$. Their total air gap length $2g$ remain unchanged as the mover moves. Therefore the flux excited by the coil and the inductance of the coil is

$$
\phi_c(i) = \mu_0\frac{2NA_ci(t)}{2g},
$$

$$
L = \frac{2N\phi_c(i)}{i}= \mu_0\frac{2N^2A_c}{g}.
$$

### 2.2 Permanent magnet excitation

When the displacement of the mover is $x$, the air gap on the left becomes $g+x$, while the right one becomes $g-x$. The two air gaps are in parallel from magnetic circuit perspective, resulting in a total magnetic reluctance of

$$
\mathcal{R}_{gt} = \frac{1}{\frac{\mu_0A_c}{g+x}+\frac{\mu_0A_c}{g-x}}=\frac{g^2-x^2}{2\mu_0A_cg}.
$$

Them MMF of the permanent magnet is

$$
\mathcal{F}_{PM} = H_ch_m.
$$

The total internal reluctance (permanent magnet and the vertical air gap $g_0$) is

$$
\mathcal{R}_i = \frac{h_m}{\mu_0\mu_rA_m}+\frac{g_0}{\mu_0A_m}.
$$

The total flux from the permanent magnet is then

$$
\phi_m(x) = \frac{\mathcal{F}}{\mathcal{R}_i+\mathcal{R}_{gt}}=\frac{\mu_0H_ch_m}{\frac{h_m}{\mu_rA_m}+\frac{g_0}{A_m}+\frac{g^2-x^2}{2A_cg}}.
$$

The permanent magnet flux linkage of the right hand coil is

$$
\lambda_R(x) = N\phi_m(x)\frac{\mathcal{R}_{gt}}{R_R} = N\frac{\mu_0H_ch_m}{\frac{h_m}{\mu_rA_m}+\frac{g_0}{A_m}+\frac{g^2-x^2}{2A_cg}}\frac{g+x}{2g} .
$$

That of the left hand coil is

$$
\lambda_L(x) = N\phi_m(x)\frac{\mathcal{R}_{gt}}{R_L} = N\frac{\mu_0H_ch_m}{\frac{h_m}{\mu_rA_m}+\frac{g_0}{A_m}+\frac{g^2-x^2}{2A_cg}}\frac{g-x}{2g} .
$$

The total permanent magnet excited flux in the coil is

$$
\begin{aligned}
\lambda_m(x) = \lambda_R(x) - \lambda_L(x) &= N\frac{\mu_0H_ch_m}{\frac{h_m}{\mu_rA_m}+\frac{g_0}{A_m}+\frac{g^2-x^2}{2A_cg}}\frac{x}{g}\\
&= \frac{\mu_0NH_ch_m}{\frac{\frac{h_m/\mu_r+g_0}{A_m}+\frac{g}{2A_c}}{x}-\frac{x}{2A_cg}}\cdot\frac{1}{g}.
\end{aligned}
$$

When $A_c$ and $A_m$ are comparable, and $x\ll h_m$, the first item of the denominator dominates, therefore

$$
\begin{aligned}
\lambda_m(x) &\approx \frac{\mu_0NH_ch_m}{\frac{\frac{h_m/\mu_r+g_0}{A_m}+\frac{g}{2A_c}}{x}}\cdot\frac{1}{g}\\
&=\frac{\mu_0NH_ch_m}{\frac{h_m/\mu_r+g_0}{A_m}+\frac{g}{2A_c}}\cdot\frac{x}{g}.
\end{aligned}
$$

Since the flux linkage excited by the current is considered to be a constant, the force equation can be solved as

$$
\begin{aligned}
f(x,i) &= i(t)\frac{\mathrm{\partial}\lambda(x,i)}{\mathrm{\partial}x} = i(t)\frac{\mathrm{\partial}\lambda_m(x)}{\mathrm{\partial}x} \\
&\approx\frac{\mu_0NH_ch_m}{\frac{h_m/\mu_r+g_0}{A_m}+\frac{g}{2A_c}}\cdot\frac{i(t)}{g},
\end{aligned}
$$
which is a linear function of current, independent from the displacement and is desirable for high precision actuators.

### 2.3 Negative stiffness

It is worth noticing that when we linearize the the force equation, the dependency of the force on the displacement is neglected. 
If we take the second derivative of the exact permanent magnet flux linkage equation, the result would be always positive, i.e., the magnetic force will increase when the displacement is getting larger. This of course will result in a unstable behaviour. This phenomenon is often addressed as "negative stiffness" in electromechanics as oppossing to the "positive stiffness" normally provided by springs or flextures.

To make the actuator stable and usable, active controller or a flexure/spring should be added to provide positive stiffness.

## 3 Numerical modelling and simulation

|Number of turns one side ($N$) | Horizontal air gap ($g$) | Vertical air gap ($g_0$) | PM height ($h_m$) | PM coercivity ($H_c$) | PM relative permeability ($\mu_r$) | PM sectional area ($A_m$) | Core sectional area ($A_c$)|
|-------------------------------|--------------------------|--------------------------|-------------------|-----------------------|------------------------------------|---------------------------|----------------------------|
|          150                  | $1~\mathrm{mm}$          | $0.5~\mathrm{mm}$        | $20~\mathrm{mm}$  |  $910~\mathrm{kA/m}$  |          1.05                      | $2\times 10^{-4}~m^2$     | $1\times 10^{-4}~m^2$      |

The code below simulates the motion of the mover when the coil is supplied with a switching voltage source.

````julia
# Required packages: PyPlot, DifferentialEquations
# Install with: import Pkg; Pkg.add(["PyCall", "PyPlot", "DifferentialEquations", "OrdinaryDiffEqLowOrderRK"])
using PyPlot
using PyCall
using DifferentialEquations
using OrdinaryDiffEqLowOrderRK

# Actuator parameters (all SI units)
N  = 35
g  = 1e-3
g0 = 0.5e-3
hm = 20e-3
Hc = 910e3
mur = 1.05
Am = 2e-4
Ac = 1e-4
mu0 = 4 * π * 1e-7

# Mass and spring constant
k   = 120e3               # spring constant of the flexure
m   = Ac * 40 / 1000 * 7870   # mass (from volume and density)

# Derived magnetic parameters
km = mu0 * N * Hc * hm / ((hm/mur + g0)/Am + g/(2*Ac)) / g
L  = 2 * N^2 * mu0 * Ac / g

# PWM function generator (works on scalars; use broadcasting for arrays)
function v_pwm(t, D, Vs, fs)
    t_scaled = t * fs
    x_period = mod(t_scaled, 1.0)   # remainder, positive (same as Python %)
    if x_period <= D
        return Vs
    else
        return -Vs
    end
end

# ODE model for the hybrid reluctance actuator (in-place form for efficiency)
function hr_actuator_ode!(du, u, p, t)
    Vs, D, fs, R, L, k, m = p
    u_pwm = v_pwm(t, D, Vs, fs)
    # System matrix A and input vector B
    A = [-R/L   -km/L   0;
          km/m   0     -k/m;
          0      1      0]
    B = [1/L; 0; 0]
    du .= A * u + B * u_pwm
end

# Initial condition
x0 = [0.0, 0.0, 0.0]

# Circuit and PWM parameters
R  = 1.0
Vs = 80.0
fs = 5000.0
n_period = 50
D = 0.55

# Time span and parameter tuple
tspan = (0.0, 1/fs * n_period)
p = (Vs, D, fs, R, L, k, m)

# Build and solve the ODE problem
prob = ODEProblem(hr_actuator_ode!, x0, tspan, p)
sol = solve(prob, RK4())

# Extract time vector and solution matrix (rows = time points, columns = states)
t = sol.t
X = hcat(sol.u...)'   # transpose so that each row corresponds to one time

# Plotting setup (matplotlib style via PyPlot)
rcParams = PyDict(matplotlib."rcParams")
rcParams["text.usetex"] = true
rcParams["figure.figsize"] = [6.0, 4.0]
rcParams["figure.dpi"] = 200

# Create subplots
fig1, axes = plt.subplots(4, 1)
ax0, ax1, ax2, ax3 = axes

# Voltage (broadcast v_pwm over the time vector)
ax0.plot(t, v_pwm.(t, D, Vs, fs))
ax0.set_ylabel("Voltage (V)")
ax0.set_xlabel("Time (s)")

# Current
ax1.plot(t, X[:, 1])
ax1.set_ylabel("Current (A)")
ax1.set_xlabel("Time (s)")

# Speed (converted to mm/s)
ax2.plot(t, 1000 * X[:, 2])
ax2.set_ylabel("Speed (mm/s)")
ax2.set_xlabel("Time (s)")

# Position (converted to mm)
ax3.plot(t, 1000 * X[:, 3])
ax3.set_ylabel("Position (mm)")
ax3.set_xlabel("Time (s)")

fig1.tight_layout()
#display(fig1)   # or show() if running in a script without a notebook
````
