# Interdisciplinary Challenge: Numerical Methods for Electrical Engineering

## Welcome

### Course syllabus

The technical content of this IC is organised in three progressive blocks that interleave a mathematics strand focused on the finite-element method with an electrical-engineering strand focused on modelling and simulation of electromechanical systems.

* Block 1 (Weeks 1–2) introduces 1-D FEM (weak form, linear elements, assembly, verification) while using simple electrical circuits—solvable by hand or symbolically—to introduce ODE modelling, state-space representation, basic solvers, and discretisation methods for digital control (ZOH, Euler, bilinear). Students leave with their own 1-D FEM code and electrical circuit simulation code.
* Block 2 (Weeks 3–4) advances to 2-D FEM on rectangular domains (mesh generation, higher-order elements, time-dependent problems). On the EE side modelling tools are applied to coreless haptic actuators and hybrid reluctance actuators, covering their operating principles, energy and force calculation, full electromechanical state-space models, and discrete-time realisations. Students leave with a working 2-D FEM code and magnetic circuit based dynamic model of the chosen actuator.
* Block 3 (Weeks 5–7) extends the FEM implementation to nonlinear materials, force extraction, and electromechanical coupling. Teams build their own modular 2-D FEA software and use it to analyse the chosen actuator, combining field solutions with continuous or discrete-time system models.
  
By the end of the course students should be able to implement the numerical methods and apply them to the modelling, analysis and digital-control preparation of electromechanical actuators.

Detailed syllabus is as follows.

**Duration:** 7 weeks × 4 lecture hours/week  

**Math focus:** FEM principles and implementation only (weak form, shape functions, assembly, solvers, verification).  
**EE focus:**  
- Block 1: Simple circuits that can be solved by hand or symbolically — introduce ODE modelling, state-space, basic solvers, and discretisation.  
- Blocks 2–3: Advance to coreless haptic actuators and hybrid reluctance actuators (principles, energy & force, full electromechanical models, digital-control discretisation).

### Block 1 (Weeks 1–2, 8 h) – 1-D FEM + friendly introduction via RL / RLC circuits

| Week | Session | Duration | Math Part (FEM) | EE Part | Notes / Deliverables |
|------|---------|----------|-----------------|---------|----------------------|
| 1 | Day 1 | 2 h | Course overview; software setup (Julia, GMSH, ParaView) | Motivation; simple RL and RLC circuits as electromechanical building blocks | Install & verify toolchain with given notebooks|
| 1 | Day 2 | 2 h | Scalar Poisson equation; introduction to weak form and linear Lagrange elements (1-D) | Hand/symbolic solution of RL and RLC circuits (SymPy); source term, solution and derivative | Compare analytical results with first numerical experiments |
| 2 | Day 1 | 2 h | 1-D mesh generation, element-by-element assembly (variable diffusion), linear solve, visualisation | State-space representation of RL / RLC circuits; basic ODE solvers and solver choices (explicit/implicit Euler, trapezoidal, RK) | Own 1-D FEM code skeleton; compare with DifferentialEquations.jl |
| 2 | Day 2 | 2 h | Verification of 1-D FEM against analytical solutions; brief non-uniform meshes | Discretisation of continuous RL / RLC models for digital control: ZOH, forward/backward Euler, bilinear (Tustin) | Complete 1-D FEM code; discrete-time circuit models |

**Block-1 assignment:**
 - Math:
 - EE: make state-space model, simulate it with own solver and compare the result with DifferentialEquations.jl for a chosen circuit between: 1) buck-boost converter, 2) sine PWM inverter with LC filtered R load.
**Block-1 outcome:** Working 1-D FEM code + solid, hand-verifiable understanding of ODE modelling, state-space, solvers and discretisation of simple circuits.

### Block 2 (Weeks 3–4, 8 h) – 2-D FEM + introduction of the two actuators

| Week | Session | Duration | Math Part (FEM) | EE Part | Notes / Deliverables |
|------|---------|----------|-----------------|---------|----------------------|
| 3 | Day 1 | 2 h | 2-D Poisson equation; manufactured solutions & separation of variables; Dirichlet/Neumann BCs | Operation principles of coreless haptic actuators and hybrid reluctance actuators; magnetic-circuit approximations and analytical field calculation | Analytical notebooks provided for verification |
| 3 | Day 2 | 2 h | Structured 2-D mesh generation, weak form, assembly, sparse solve, post-processing (Plots.jl + ParaView/VTK) | Energy and force in the two actuators; reduction of electromechanical systems to ODEs | Own 2-D rectangular FEM code |
| 4 | Day 1 | 2 h | Higher-order (quadratic) elements; convergence studies; material interfaces | Full state-space models of the electromechanical actuators; solver selection criteria (stiffness, accuracy, real-time constraints) | Error tables and refined meshes |
| 4 | Day 2 | 2 h | Time-dependent 2-D problems via method of lines; coupling spatial FEM to DifferentialEquations.jl | Discretisation for digital control (ZOH, Euler, bilinear) applied to the actuator state-space models; sampling effects | 2-D FEM + time integration; discrete-time actuator models |

**Block-2 assignment:**
 - Math:
 - EE: Derive state-space model of the chosen actuator based on magnetic circuit/analytical method, simulate its dynamics when fed by PWM voltage source and compare the results with DifferentialEquations.jl and the discretized model
**Block-2 outcome:** Working 2-D FEM code on rectangular domains + continuous and discrete state-space models of the two actuators, grounded in the simple-circuit methods of Block 1.

### Block 3 (Weeks 5–7, 12 h) – Coupled 2-D FEM for actuators + team FEA software

| Week | Session | Duration | Math Part (FEM) | EE Part | Notes / Deliverables |
|------|---------|----------|-----------------|---------|----------------------|
| 5 | Day 1 | 2 h | Nonlinear FEM (Newton–Raphson) for magnetic saturation; element material updates | Nonlinear energy and force in saturated hybrid reluctance actuators; impact on the ODE / state-space models | Extend 2-D code to nonlinear magnetostatics |
| 5 | Day 2 | 2 h | Force extraction (virtual work / Maxwell stress) from FEM solutions | Coreless haptic actuator force characteristics; comparison of analytical, ODE, and FEM predictions | Force–displacement curves at varying current levels |
| 6 | Day 1 | 2 h | Motion handling (moving mesh or fixed mesh + remapping); coupling FEM to rigid-body dynamics | Electromechanical system simulation: coupling field solver to mechanical ODE / state-space model | Team software architecture design |
| 6 | Day 2 | 2 h | Modular code structure, GMSH geometry import, VTK export, testing practices | Digital-control discretisation (ZOH, Euler, bilinear) of the fully coupled electromechanical model; real-time considerations | Begin team implementation |
| 7 | Day 1 | 2 h | Verification, validation, and documentation of a small FEA code | Case studies: optimising force, stroke, or bandwidth of hybrid reluctance and coreless haptic actuators | Team coding + debugging |
| 7 | Day 2 | 2 h | Final code review and performance considerations | Team demonstrations of the developed FEA software analysing the chosen actuator | Team report submission |

**Block-3 outcome:** Teams deliver a modular 2-D FEA software package that can analyse hybrid reluctance or coreless haptic actuators, including nonlinear magnetics, force computation, electromechanical coupling, and discrete-time models suitable for digital control.


### Course evaluation

The technical component of the course is assessed through a single team report submitted at the end of the course. The report must be prepared as an incremental Jupyter or Pluto notebook that documents all implementations and solutions developed across the three blocks, together with the required reflections on soft skills. The final grade is determined exclusively by this final report. Intermediate feedback is given after each block for formative purposes only and does not contribute to the grade.
